# Part D: Gaussian Mixture Models


```python
import time

from sklearn.datasets import load_digits
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import (
    adjusted_rand_score,
    normalized_mutual_info_score,
    silhouette_score,
)
digits = load_digits()
X = digits.data
y_true = digits.target
```

```python
#kmeans
model_kmeans = KMeans(
    n_clusters=10,
    n_init=10,
    random_state=47,
)

start = time.perf_counter()
labels_kmeans = model_kmeans.fit_predict(X)
runtime_kmeans = time.perf_counter() - start

print(
    f"ARI={adjusted_rand_score(y_true, labels_kmeans):.4f}, "
    f"NMI={normalized_mutual_info_score(y_true, labels_kmeans):.4f}, "
    f"Silhouette={silhouette_score(X, labels_kmeans):.4f}, "
    f"Runtime={runtime_kmeans:.4f}s"
)
```

```python
#GMM diagonal 每个成分只学习各特征的方差，不建模特征之间的相关性
models_diag = {}

for n in [5, 10, 15]:
    model_diag = GaussianMixture(
        n_components=n,
        covariance_type="diag",
        n_init=3,
        max_iter=200,
        random_state=47,
    )

    start = time.perf_counter()
    labels_diag = model_diag.fit_predict(X)
    runtime_diag = time.perf_counter() - start

    models_diag[n] = model_diag

    print(
        f"GMM diag, components={n}, "
        f"ARI={adjusted_rand_score(y_true, labels_diag):.4f}, "
        f"NMI={normalized_mutual_info_score(y_true, labels_diag):.4f}, "
        f"Silhouette={silhouette_score(X, labels_diag):.4f}, "
        f"Runtime={runtime_diag:.4f}s"
    )

for n in [5, 10, 15]:
    model = models_diag[n]

    print(f"\nGMM diag, components={n}")

    print("Mixture weights:")
    print(model.weights_)

    print("Component means:")
    print(model.means_)

    print("Covariance parameters:")
    print(model.covariances_)

    print("Predicted cluster probabilities:")
    print(model.predict_proba(X[:3]))
```

```python
#GMM full 每个成分既学习各特征的方差，也学习特征之间的相关性
models_full = {}
for n in [5, 10, 15]:
    model_full = GaussianMixture(
        n_components=n,
        covariance_type="full",
        n_init=3,
        max_iter=200,
        random_state=47,
    )

    start = time.perf_counter()
    labels_full = model_full.fit_predict(X)
    runtime_full = time.perf_counter() - start

    models_full[n] = model_full

    print(
        f"GMM full, components={n}, "
        f"ARI={adjusted_rand_score(y_true, labels_full):.4f}, "
        f"NMI={normalized_mutual_info_score(y_true, labels_full):.4f}, "
        f"Silhouette={silhouette_score(X, labels_full):.4f}, "
        f"Runtime={runtime_full:.4f}s"
    )

for n in [5, 10, 15]:
    model = models_full[n]

    print(f"\nGMM full, components={n}")

    print("Mixture weights:")
    print(model.weights_)

    print("Component means:")
    print(model.means_)

    print("Covariance parameters:")
    print(model.covariances_)

    print("Predicted cluster probabilities:")
    print(model.predict_proba(X[:3]))
```

What is the difference between hard clustering and soft clustering?

Ans: Hard clustering assigns each sample to exactly one cluster. Soft clustering gives each sample a probability of belonging to each cluster.

How is a GMM different from K-means?

Ans: K-means assigns each point to one cluster based on distance. A GMM gives each point a probability of belonging to each cluster and can model different cluster shapes.

What assumptions does a Gaussian component make about the data?

Ans: It assumes its data follows a normal distribution around a mean.
