## Hands-on Task 1: Python and NumPy


```python
import numpy as np
import pandas as pd
import torch
```

```python
np.random.seed(47)
matrix = np.random.rand(1000,20)
```

```python
#Calculate the mean of every feature
mean = np.mean(matrix, axis=0)
print("mean: ",mean)
```

```python
#Calculate the standard deviation of every feature
std = np.std(matrix, axis=0)
print("standard deviation: ",std)
```

```python
#Normalize every feature
normal_features = (matrix - mean) / std
print("normalized features: ",normal_features)
```

```python
#Find the five samples with the largest Euclidean norm(the length of the vector)
euclidean_norms = np.argsort(np.linalg.norm(matrix, axis=1))[-5:]

print("five samples with the largest Euclidean norm: ",euclidean_norms)
```

```python
#Calculate the pairwise Euclidean distance between the first 20 samples(the length between each pair of samples)
def euclidean_distance(x, y):
    return np.sqrt(np.sum((x - y) ** 2, axis=-1))

pairwise_euclidean = euclidean_distance(
    matrix[:20][:,None,:],#take one sample
    matrix[:20][None,:,:]#take all 20 samples
)

print("Pairwise Euclidean distance:\n", pairwise_euclidean)
```

```python
#Calculate the pairwise cosine similarity between the first 20 samples.(similarity between two vectors' direction)
def cosine_similarity(x, y):
    dot_product = np.sum(x * y, axis=-1)
    norm_x = np.linalg.norm(x, axis=-1)
    norm_y = np.linalg.norm(y, axis=-1)
    return dot_product / (norm_x * norm_y)

pairwise_cosine = cosine_similarity(
    matrix[:20][:,None,:],
    matrix[:20][None,:,:]
)

print("Pairwise Cosine similarity:\n", pairwise_cosine)
```
